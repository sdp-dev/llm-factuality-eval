```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("allen-institute-for-ai/CORD-19-research-challenge")

print("Path to dataset files:", path)
```

```python
import os
```

```python
# os.listdir(path+'/document_parses/pmc_json')
```

```python
path_pmc = path+'/document_parses/'
```

```python
from pandas import json_normalize

df_pmc=pd.DataFrame()

count=0

for file_path in os.listdir(path+'/document_parses/pmc_json/'):
  count+=1
  if count>100000:
    break

  with open(path+'/document_parses/pmc_json/'+file_path, 'r') as file:
    data = json.load(file)

    temp = json_normalize(data, record_path='body_text', meta=['metadata','paper_id'])
    temp['paper_text'] = "".join(temp['text'])
    to_compare = ''.join(temp['paper_text']).replace(' ','')
    if "sideeffects" in to_compare or "adverseeffects" in to_compare or "complications" in to_compare or "symptoms" in to_compare:
      print('doc:',count)
      df_pmc = pd.concat([df_pmc, temp[['paper_id','paper_text']].drop_duplicates()], axis=0).reset_index(drop=True)
```

```python
df_pmc.to_csv('filtered_dataset_new.csv', index=False)
```
